```python
# 프로젝트 위치가 GummyBearing인지 확인

%load_ext autoreload
%autoreload 2

from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()

if not (PROJECT_ROOT / "src").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent

if not (PROJECT_ROOT / "src").is_dir():
    raise FileNotFoundError("프로젝트의 src 폴더를 찾을 수 없습니다.")

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.ims_features import extract_ims_features

print("프로젝트 위치:", PROJECT_ROOT)
```

```python
# 함수를 불러오고 데이터 경로를 확인하고 데이터가 존재하는지 확인

from src.ims_features import extract_ims_features

DATA_ROOT = PROJECT_ROOT / "DATA" / "IMS_Bearing_Dataset"

test_folders = {
    1: DATA_ROOT / "1st_test",
    2: DATA_ROOT / "2nd_test",
    3: DATA_ROOT / "3rd_test",
}

for test_number, folder in test_folders.items():
    print(f"test {test_number}: {folder.is_dir()} | {folder}")
```

```python
# 특징 계산 실행

results = {}
error_results = {}

for test_number, folder in test_folders.items():
    features_df, errors_df = extract_ims_features(
        folder_path=folder,
        test_number=test_number,
    )

    results[test_number] = features_df
    error_results[test_number] = errors_df

    print(
        f"test {test_number}: "
        f"결과 {features_df.shape}, "
        f"실패 파일 {len(errors_df)}개\n"
    )
```

```python
# 결과 확인

from IPython.display import display

features_1st = results[1]
features_2nd = results[2]
features_3rd = results[3]

print("1st 결과")
display(features_1st.head(8))

print("2nd 결과")
display(features_2nd.head(4))

print("3rd 결과")
display(features_3rd.head(4))
```

```python
# 오류와 결측값 확인

feature_columns = ["rms", "kurtosis", "peak", "crest_factor"]

for test_number in test_folders:
    print(f"\n===== test {test_number} =====")

    print("처리 실패 파일 수:", len(error_results[test_number]))

    if not error_results[test_number].empty:
        display(error_results[test_number])

    print("계산된 특징의 NaN 개수:")
    print(results[test_number][feature_columns].isna().sum())
```

```python
# csv 파일 저장

OUTPUT_ROOT = PROJECT_ROOT / "DATA" / "features"
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

test_names = {
    1: "1st_test",
    2: "2nd_test",
    3: "3rd_test",
}

for test_number, test_name in test_names.items():
    results[test_number].to_csv(
        OUTPUT_ROOT / f"{test_name}_features.csv",
        index=False,
        encoding="utf-8-sig",
    )

    error_results[test_number].to_csv(
        OUTPUT_ROOT / f"{test_name}_errors.csv",
        index=False,
        encoding="utf-8-sig",
    )

print("저장 완료:", OUTPUT_ROOT)
```
